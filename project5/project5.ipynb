Целевая аудитория приложения — начинающие инвесторы. Получить доступ к финансовым рынкам можно со смартфона либо через веб-версию.

Появилась гипотеза о недостаточной финансовой грамотности пользователей: они не понимают, как работать с активами разной степени риска. Клиенты покупают активы с высоким риском и теряют деньги, что снижает вовлечённость.
Команда продукта решила обновить онбординг пользователей и добавить в него детальную информацию о различиях финансовых активов и связанных с ними рисках. Однако появилось опасение, что углублённый онбординг отпугнёт клиентов от пополнения депозита и более рискованных вложений. 

Чтобы разобраться, команда решила провести A/B-эксперимент. 

### А/В-эксперимент

В эксперименте участвовали новые пользователи, которые зарегистрировались в приложении со 2 по 15 июня 2025 года. 
Пользователей случайным образом разделили на две равные группы: 
- Контрольная группа проходила стандартный онбординг без обязательного обучения.
- Тестовая группа проходила обновлённый онбординг с подробной информацией об активах и связанных с ними рисках. 

После разделения активность пользователей анализировали в течение недели.

Цель эксперимента — оценить, как обновлённый онбординг влияет на поведение пользователей и их дальнейшую инвестиционную активность.

При внедрении новой фичи команда продукта выдвинула такие гипотезы:
- Гипотеза роста: обучающий онбординг помогает пользователям лучше понимать принципы инвестирования, поэтому они будут чаще открывать второй депозит.
- Гипотеза риска: информация о возможных потерях и высоких рисках отпугнёт некоторых новичков, особенно самых осторожных, что снизит конверсию в первый депозит.
- Дополнительная гипотеза: после нового онбординга пользователи, которые выбрали высокорискованные активы, будут чаще, чем раньше, возвращаться и открывать второй депозит. При старом онбординге пользователи часто покупали активы с высоким риском без понимания последствий. Это приводило к потерям и оттоку после первого депозита. 

Чтобы увидеть полную картину, недостаточно следить за метриками конверсии. Команда выбрала для эксперимента такой набор метрик: 
- Ключевая метрика — средняя сумма всех депозитов на одного пользователя (включая тех, кто установил приложение или открыл веб-версию).
- Барьерная метрика — конверсия из регистрации в первый депозит.
- Вспомогательная метрика 1 — конверсия из первого депозита во второй.
- Вспомогательная метрика 2 — средняя сумма всех депозитов на пользователя, который открыл хотя бы один депозит.

Ожидалось, что ключевая и барьерная метрики не упадут, а вспомогательные покажут значительный рост.

**Задача 2.2.** Анализ воронок событий.
- Постройте классическую и пошаговую воронки;
- Охарактеризуйте пользовательский путь в приложении и определите узкие места воронки. 


````python
# Задаем порядок шагов в воронке
funnel_steps = ['install / open_web', 'introduction', 'registration', 'main_page', 'onboarding_complete', 'first_deposit', 'asset_purchase', 'second_deposit']

# Фильтруем датасет оставляя только нужные шаги воронки
funnel_df_hist_steps = df_hist[df_hist['event_name'].isin(funnel_steps)]
# Считаем число уникальных пользователей на каждом шаге
funnel_counts = funnel_df_hist_steps.groupby('event_name')['user_id'].nunique().reindex(funnel_steps)

# Собираем таблицу воронки
funnel_df_hist = pd.DataFrame({
   'step': funnel_counts.index,
   'users': funnel_counts.values
})

# Добавляем конверсии и округляем до 1 знака после запятой
funnel_df_hist['conversion_from_first_%'] = round(funnel_df_hist['users'] / funnel_df_hist['users'].iloc[0] * 100,1)
funnel_df_hist['conversion_from_previous_%'] =round(funnel_df_hist['users'] / funnel_df_hist['users'].shift(1) * 100,1)

display(funnel_df_hist)
````

````python
# Визуализируем данные
plt.figure(figsize=(12, 5))
# Строим barplot
# Cтроим простой график воронки

bars=plt.bar(funnel_df_hist['step'], funnel_df_hist['conversion_from_first_%'], color='skyblue')
plt.xticks(rotation=45)
plt.title('Классическая воронка: конверсия от первого шага')
plt.ylabel('Конверсия от первого шага (%)')
plt.xlabel('Этапы воронки')

# Подписи столбцов
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # позиция по оси X (центр столбца)
        height + 1,  # позиция по оси Y (чуть выше вершины столбца)
        f'{height:.1f}%',  # текст подписи (значение с одним знаком после запятой + %)
        ha='center',  # выравнивание по горизонтали: по центру
        va='bottom'   # выравнивание по вертикали: снизу
    )
plt.show()

# Визуализируем данные, кроме первого шага, чтобы не выводить NaN
plt.figure(figsize=(12, 5))
# Строим barplot
# Cтроим простой график воронки

bars=plt.bar(funnel_df_hist['step'][1:], funnel_df_hist['conversion_from_previous_%'][1:], color='skyblue')
plt.xticks(rotation=45)
plt.title('Step-by-step воронка: конверсия от предыдущего шага')
plt.ylabel('Конверсия от предыдущего шага (%)')
plt.xlabel('Этапы воронки')

# Подписи столбцов
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # позиция по оси X (центр столбца)
        height + 1,  # позиция по оси Y (чуть выше вершины столбца)
        f'{height:.1f}%',  # текст подписи (значение с одним знаком после запятой + %)
        ha='center',  # выравнивание по горизонтали: по центру
        va='bottom'   # выравнивание по вертикали: снизу
    )
plt.show()
````

Пользовактельский путь начинается с установки приложения или/и открытие веб‑версии. После чеого проходит вводное знакомство, регистрацию. После регистрации пользователь попадает на главный экран, следующий шаг - финальный этап знакомства с приложением. Далее пользователь совершает первый депозит (пополняет счет в приложении), после чего совершается первая инвестиционная операция - покупка активов, и далее пользователь может заного пополнить депозит.

По классической воронке из 100% пользователей, начавших путь с установки приложения, до первого пополнения депозита доходят 67.5% пользователей, до покупки активов - 37.5%, а до второго пополнения счета - 13.2%. 
Наибольшие потери на шагах: первый депозит-покупка активов - 37.5% (конверсия от предыдущего шага 55.6%), и покупка актива-второй депозит - 13.2% (конверсия от предыдущего шага 35.1%). Именно эти два этапа является узкими местами воронки.

**Задача 3.2.** Сравнение воронок событий.
- Сравните пользовательский путь новых пользователей в тестовой и контрольной группах.
- Постройте классическую и последовательную воронки, выполните визуализацию и сделайте промежуточный вывод. 

````python
# Задаем порядок шагов в воронке
funnel_steps = ['install / open_web', 'introduction', 'registration', 'main_page', 'onboarding_complete', 'first_deposit', 'asset_purchase', 'second_deposit']
results = []
for group in df['group'].unique():
    group_df=df[df['group']==group]
    # Фильтруем датасет оставляя только нужные шаги воронки
    funnel_df_steps = group_df[group_df['event_name'].isin(funnel_steps)]
    # Считаем число уникальных пользователей на каждом шаге
    funnel_counts = funnel_df_steps.groupby('event_name')['user_id'].nunique().reindex(funnel_steps)
   
    for step, users in funnel_counts.items():
        results.append({
             'step': step,
            'users_control': users,
            'group': group
        })

# Преобразуем список словарей в DataFrame
results_df = pd.DataFrame(results)

# Рассчитываем конверсии после создания DataFrame
# Конверсия от первого шага воронки
first_step_users = results_df.groupby('group')['users_control'].transform('first')
results_df['conversion_from_first_%'] = round(results_df['users_control'] / first_step_users * 100, 1)


# Рассчитываем конверсию с учётом группы
results_df['conversion_from_previous_%'] = (
    results_df['users_control'] /
    results_df.groupby('group')['users_control'].shift(1) * 100
).round(1)


# выводим
display(results_df)
````

````python
# Визуализируем данные
plt.figure(figsize=(12, 5))
# Строим barplot
# Cтроим простой график воронки

# Строим barplot
bars=sns.barplot(data=results_df, x='step', y='conversion_from_first_%', hue='group')

# Добавляем подписи для каждого столбца
for bar_container in bars.containers:
    for bar_instance in bar_container:
        height = bar_instance.get_height()
        plt.text(
            bar_instance.get_x() + bar_instance.get_width() / 2,
            height,
            f'{height:.1f}%',
            ha='center',
            va='bottom'
        )

plt.xticks(rotation=45)
plt.title('Классическая воронка: конверсия от первого шага')
plt.ylabel('Конверсия от первого шага (%)')
plt.xlabel('Этапы воронки')

plt.show()

# Визуализируем данные, кроме первого шага, чтобы не выводить NaN
plt.figure(figsize=(12, 5))

# Фильтруем данные: исключаем первый шаг воронки
filtered_df = results_df[results_df['step'] != funnel_steps[0]]

# Строим barplot на отфильтрованных данных
bars = sns.barplot(data=filtered_df, x='step', y='conversion_from_previous_%', hue='group')

# Добавляем подписи для каждого столбца
for bar_container in bars.containers:
    for bar_instance in bar_container:
        height = bar_instance.get_height()
        plt.text(
            bar_instance.get_x() + bar_instance.get_width() / 2,
            height,
            f'{height:.1f}%',
            ha='center',
            va='bottom'
        )

plt.xticks(rotation=45)
plt.title('Step-by-step воронка: конверсия от предыдущего шага (без первого шага)')
plt.ylabel('Конверсия от предыдущего шага (%)')
plt.xlabel('Этапы воронки')

plt.show()
 
````

Гипотеза ключевой метрики:

- H0: средняя сумма всех депозитов на одного пользователя в тестовой и контролькой группах одинакова.
- H1: средняя сумма всех депозитов на одного пользователя в тестовой и контрольной группах различается.

````python
# Сначала считаем сумму депозитов для каждого пользователя
user_deposits = df.groupby('user_id')['amount'].sum().reset_index()
user_deposits.rename(columns={'amount': 'total_deposit'}, inplace=True)

# Затем присоединяем обратно к исходному датафрейму, чтобы сохранить информацию о группах
df_with_deposits = pd.merge(df, user_deposits, on='user_id', how='left')
df_with_deposits=df_with_deposits[['user_id', 'group', 'total_deposit']].drop_duplicates()
groups = sorted(df_with_deposits['group'].unique())
group_data = {group: df_with_deposits[df_with_deposits['group'] == group]['total_deposit'] for group in groups}
# Создаём список для хранения результатов
results = []

# Проводим попарные сравнения всех групп
for i in range(len(groups)):
    for j in range(i + 1, len(groups)):
        group1 = groups[i]
        group2 = groups[j]
        _, p_value = ttest_ind(group_data[group1], group_data[group2], equal_var=False)

        # Рассчитываем средние значения
        mean_a = group_data[group1].mean()
        mean_b = group_data[group2].mean()

        # Рассчитываем абсолютный и относительный lift
        absolute_lift = abs(mean_a - mean_b)
        relative_lift_percent = group_data[group2].mean() / group_data[group1].mean() - 1

        # Добавляем результаты в список
        results.append({
            'Group A': group1,
            'Group B': group2,
            'p-value': p_value,
            'Mean A': mean_a,
            'Mean B': mean_b,
            'Absolute Lift': absolute_lift,
            'Relative Lift (%)': round(100*relative_lift_percent, 2)
        })

# Создаём DataFrame из результатов
results_df = pd.DataFrame(results)
print(results_df.head())
````

Средняя сумма всех депозитов на одного пользователя в контрольной группе составила 90,2, в тестовой — 92,4. Абсолютное изменение составило 2,2 единицы, относительный прирост — около 2,46%.

Проанализируем барьерную метрику - конверсия из регистрации в первый депозит:
Гипотеза:
- Н0: конверсия из регистрации в первый депозитв контрольной и тестовых группах одинакова.
- Н1: конверсия из регистрации в первый депозитв контрольной и тестовых группах различна.

````python
# Рассчитываем числитель и знаменатель метрики
df_test = df.groupby(['user_id', 'group']).agg(
    total_rides=('event_name', lambda x: x.eq('registration').nunique()),
    successful_rides=('event_name', lambda x: x.eq('first_deposit').nunique())
).reset_index()


# Получаем список всех уникальных групп с сортировкой
groups = sorted(df_test['group'].unique())

# Создаём словарь для хранения данных по группам
group_data = {}
for group in groups:
    group_data[group] = df_test[df_test['group'] == group]

# Создаём список для хранения результатов
results = []

# Проводим попарные сравнения всех групп
for i in range(len(groups)):
    for j in range(i + 1, len(groups)):

        group1 = groups[i]
        group2 = groups[j]

        # Расчет коэффициента линеаризации
        ratio = (
            group_data[group1]['successful_rides'].sum()
            / group_data[group1]['total_rides'].sum()
        )

        # Линеаризованные значения
        lin_metric1 = (
            group_data[group1]['successful_rides']
            - ratio * group_data[group1]['total_rides']
        )

        lin_metric2 = (
            group_data[group2]['successful_rides']
            - ratio * group_data[group2]['total_rides']
        )
        _, p_value = ttest_ind(
            lin_metric1,
            lin_metric2)

        # Средние значения нелинеаризованной метрики
        mean1 = group_data[group1]['successful_rides'].sum()/ group_data[group1]['total_rides'].sum()
        mean2 = group_data[group2]['successful_rides'].sum()/ group_data[group2]['total_rides'].sum()

        # Добавление результатов
        results.append({
            'Group A': group1,
            'Group B': group2,
            'p-value': p_value,
            'Mean A': mean1,
            'Mean B': mean2,
            'Absolute Lift': mean2 - mean1,
            'Relative Lift (%)': 100 * (mean2 / mean1 - 1)
        })

# Создаём DataFrame из результатов
results_df = pd.DataFrame(results)

# Выводим результаты
print("Результаты сравнения групп:")
print(results_df)
````

По результатам эксперимента видно, что барьерная метрика в тестовой группе по сравнению с контрольной снизилась, однако разлечие статистически незначимо (p-value больше 0.05) . Конверсия из регистрации в первый депозит в тестовой группе уменьшилась на 0,8% по сравнению с контрольной группой. Следовательно, оснований считать, что новая фича повлияла на данную метрику, нет.
