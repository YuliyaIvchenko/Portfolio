Построим тепловую карту Retention Rate по когортам активированных пользователей.

````python
cmap = mcolors.LinearSegmentedColormap.from_list('green_red', ['#FF3333', '#FFFF66', '#4C9900'], N=100)

# Переводим даты в месяцы
pa_marketplace_events['event_month'] = pa_marketplace_events['event_month'].dt.to_period('M')

# Находим первое событие каждого пользователя
first_event = pa_marketplace_events.groupby('user_id')['event_month'].min().reset_index()
first_event.columns = ['user_id', 'cohort_month']

# Добавляем когорту к каждому событию пользователя
pa_marketplace = pa_marketplace_events.merge(first_event, on='user_id',  how='left')

# Считаем номер периода относительно первого события
pa_marketplace['n_period'] = pa_marketplace['event_month'].astype(int)-pa_marketplace['cohort_month'].astype(int)

# Количество активных пользователей в каждом периоде
retention = pa_marketplace.groupby(['cohort_month', 'n_period'])['user_id'].nunique().reset_index()

# Преобразуем в матрицу
retention_pivot = retention.pivot(
    index='cohort_month',
    columns='n_period',
    values='user_id'
)

# Размер когорты = количество пользователей в период 0
cohort_size = retention_pivot[0]

# Retention Rate
retention_rate = retention_pivot.div(cohort_size, axis=0) * 100

# Тепловая карта
plt.figure(figsize=(12, 6))

sns.heatmap(
    retention_rate,
    cmap=cmap,
    annot=True,
    fmt='.0f',
    linewidth=0.5
)

plt.title('Retention Rate по когортам активированных пользователей')
plt.xlabel('Период после первого события')
plt.ylabel('Когорта')
plt.show()

````

Наибольшая доля пользователей возвращается в продукт в первый месяц после первого взаимодействия (34–55% в зависимости от когорты). В дальнейшем Retention закономерно снижается: ко второму месяцу удержание составляет около 17–25%, к третьему — около 8–11%, а после шестого месяца возвращается лишь небольшая доля пользователей. Более поздние когорты демонстрируют несколько более низкое удержание уже в первом месяце, что может свидетельствовать о снижении качества привлечения пользователей.

При расчете LTV  учитывались пользователи, присутствующие одновременно в талбицах пользователей и заказов, чтобы корректно определить срок 

Расчитаем в разрезе месяца среднюю стоимость привлечения пользователя (CAC), значение LTV и ROI. И визуализируем метрики.


````python

profiles['cohort_month'] = profiles['cohort_month'].dt.to_period('M')

# Посчитаем CAC для каждого месяца
avg_CAC=profiles.groupby('cohort_month')['cac'].mean().reset_index()

# Суммируем revenue для каждого месяца
orders_LTV=orders.groupby(['cohort_month', 'n_month'])['comission'].sum().reset_index()
# количество пользователей в когорте
cohort_size = profiles.groupby('cohort_month')['user_id'].nunique().reset_index(name='user_count')
# добавим в таблицу
orders_LTV = orders_LTV.merge(cohort_size, on='cohort_month', how='left')
# кумулятивная выручка
orders_LTV['cum']=orders_LTV.groupby('cohort_month')['comission'].cumsum()
# Посчитаем LTV для каждого месяца
orders_LTV['LTV']=orders_LTV['cum']/orders_LTV['user_count']

# Посчитаем ROI для каждого месяца
orders_LTV_CAC=orders_LTV.merge(avg_CAC, on='cohort_month', how='left')
orders_LTV_CAC['ROI']=(orders_LTV_CAC['LTV']/orders_LTV_CAC['cac']-1)*100
orders_LTV_CAC=orders_LTV_CAC.sort_values(by='cohort_month')

display(orders_LTV_CAC)
````

````python
ltv_avg = orders_LTV_CAC.groupby('n_month')[['LTV', 'ROI']].mean().reset_index()

plt.figure(figsize=(12,6))

plt.plot(ltv_avg['n_month'], ltv_avg['LTV'],  marker='o')

plt.xlabel('Месяц жизни пользователя')
plt.ylabel('LTV')
plt.title('Средний LTV по месяцам жизни')
plt.grid(True)
plt.show()

plt.figure(figsize=(12,6))

plt.plot(ltv_avg['n_month'], ltv_avg['ROI'], marker='o')

plt.xlabel('Месяц жизни пользователя')
plt.ylabel('ROI (%)')
plt.title('Средний ROI по месяцам жизни')
plt.grid(True)
plt.show()

ltv_month = orders_LTV_CAC.groupby('cohort_month').agg(LTV=('LTV', 'max'), ROI=('ROI', 'max'), CAC=('cac', 'mean')).reset_index()

plt.figure(figsize=(12,6))

plt.plot(ltv_month['cohort_month'].astype(str), ltv_month['CAC'], marker='o', label='CAC')
plt.plot(ltv_month['cohort_month'].astype(str), ltv_month['LTV'], marker='o', label='LTV')

plt.xticks(rotation=45)
plt.grid(True)
plt.legend()
plt.title('Средние значения CAC и итогового LTV по когортам')
plt.show()
````

Вывод по графику «Средний LTV по месяцам жизни»
LTV закономерно растет с увеличением срока жизни пользователя. Наиболее интенсивный рост наблюдается в первые 3–4 месяца после привлечения. После 6–8 месяцев рост LTV существенно замедляется, что говорит о постепенном насыщении дохода от пользователя. Небольшое снижение на последних месяцах, вероятнее всего, связано с тем, что для этих периодов доступно меньше когорт, поэтому среднее значение становится менее устойчивым.
Вывод по графику «Средний ROI по месяцам жизни»
В первый месяц ROI отрицательный, поскольку расходы на привлечение превышают полученный доход. Уже примерно со второго месяца ROI становится положительным, что говорит о начале окупаемости привлеченных пользователей. Далее ROI стабильно увеличивается и достигает максимального значения примерно к восьмому месяцу жизни пользователей. Это свидетельствует о том, что пользователи продолжают приносить прибыль спустя несколько месяцев после привлечения.
Вывод по графику «CAC и итоговый LTV по когортам»
Стоимость привлечения пользователя (CAC) в течение года практически не меняется и остается стабильной. Для когорт, привлеченных в первой половине года, итоговый LTV значительно превышает CAC, что говорит об их высокой прибыльности. Для когорт второй половины года наблюдается снижение итогового LTV. Однако данный результат связан с тем, что поздние когорты имели меньше времени для накопления выручки, поэтому их итоговый LTV пока нельзя напрямую сравнивать с ранними когортами.


````python
# Определяем, окупилась ли когорта к данному месяцу жизни
orders_LTV_CAC['payback'] = orders_LTV_CAC['LTV'] >= orders_LTV_CAC['cac']

# Для каждой когорты ищем первый месяц окупаемости
payback = orders_LTV_CAC[orders_LTV_CAC['payback']].groupby('cohort_month')['n_month'].min().reset_index(name='payback_period')

display(payback)

plt.figure(figsize=(12,6))

plt.bar(payback['cohort_month'].astype(str), payback['payback_period'])

plt.xticks(rotation=45)
plt.ylabel('Месяцы')
plt.xlabel('Когорта')
plt.title('Срок окупаемости (Payback Period)')
plt.grid(axis='y')

plt.show()
````

Когорты до июля окупаются уже в течение первого месяца жизни. Для июльской когорты срок окупаемости увеличился примерно до двух месяцев.
Августовская когорта окупилась примерно за три месяца. Наиболее поздние когорты за рассматриваемый период не окупились. Это может быть связано как с недостаточным временем наблюдения, так и со снижением эффективности привлечения пользователей во второй половине года.

````python
mau_seg['event_month_ts'] = mau_seg['event_month'].dt.to_timestamp() 
mau_seg['cohort_month_ts'] = mau_seg['cohort_month']
# Разница в днях 
days_diff = (mau_seg['event_month_ts'] - mau_seg['cohort_month_ts']).dt.days
# Фильтруем: разница ≤ 30 дней (примерно 1 месяц) 
retention_data = mau_seg[days_diff <= 30] 
# Подсчёт активных пользователей
active_users = retention_data.groupby(['event_month', 'acq_channel'])['user_id'].nunique().reset_index()
# Общее число пользователей в когорте
total_users = mau_seg.groupby(['event_month', 'acq_channel'])['user_id'].nunique().reset_index()
# Удержание = активные / общее × 100%
user = active_users.merge(total_users, on=['event_month', 'acq_channel'], how='left')
user['retention_rate']=(user['user_id_x'] / user['user_id_y']) * 100

 # Визуализация всех метрик на одном графике
plt.figure(figsize=(14, 8))
# Цикл для построения графиков по каналам
for channel in user['acq_channel'].unique():
    df_plot_user = user[user['acq_channel'] == channel]
    plt.plot(df_plot_user['event_month'].dt.to_timestamp(), df_plot_user['retention_rate'], label=f'{channel}')


plt.title('Скользящее удержание на 30ый день по каналу привлечения')
plt.xlabel('Дата')
plt.ylabel('Удержание %')
plt.legend()
plt.grid(True)
plt.show()
````

Удержание пользователей практически одинаково для всех каналов привлечения. Существенных различий между каналами не наблюдается.

Следовательно, проблема TikTok связана не с удержанием пользователей, а с качеством привлечённой аудитории.

````python
pa_marketplace_events_users=pa_marketplace_events.merge(pa_marketplace_users, on='user_id', how='left')
count_events=pa_marketplace_events_users.groupby(['event_week', 'acq_channel'])['user_id'].nunique().reset_index()
count_orders=pa_marketplace_orders_users.groupby(['order_week', 'acq_channel'])['user_id'].nunique().reset_index()

user_conv = count_orders.merge(count_events, left_on=['order_week', 'acq_channel'], right_on=['event_week', 'acq_channel'] , how='outer')
# Конверсия = покупатели / общее число × 100%
user_conv['convertion']=(user_conv['user_id_x'] /user_conv['user_id_y']) * 100

 # Визуализация всех метрик на одном графике
plt.figure(figsize=(14, 8))
# Цикл для построения графиков по каналам
for channel in user_conv['acq_channel'].unique():
    df_plot_user = user_conv[user_conv['acq_channel'] == channel]
    plt.plot(df_plot_user['event_week'], df_plot_user['convertion'], label=f'{channel}')


plt.title('Конверсию в покупку за месяц по каналам привлечения')
plt.xlabel('Дата')
plt.ylabel('Конверсия %')
plt.grid(True)
plt.legend()
plt.show()
````

Расчитаем LTV, ROI и CAC по каналам привлечения.

````python

# Посчитаем CAC для каждого месяца
avg_CAC=profiles.groupby(['cohort_month', 'acq_channel'])['cac'].mean().reset_index()

# Суммируем revenue для каждого месяца
orders_LTV=orders.groupby(['cohort_month', 'acq_channel', 'n_month'])['comission'].sum().reset_index()
# количество пользователей в когорте
cohort_size = profiles.groupby(['cohort_month', 'acq_channel'])['user_id'].nunique().reset_index(name='user_count')
# добавим в таблицу
orders_LTV = orders_LTV.merge(cohort_size, on=['cohort_month', 'acq_channel'], how='left')
# кумулятивная выручка
orders_LTV['cum']=orders_LTV.groupby(['cohort_month', 'acq_channel'])['comission'].cumsum()
# Посчитаем LTV для каждого месяца
orders_LTV['LTV']=orders_LTV['cum']/orders_LTV['user_count']

# Посчитаем ROI для каждого месяца
orders_LTV_CAC=orders_LTV.merge(avg_CAC, on=['cohort_month', 'acq_channel'], how='left')
orders_LTV_CAC['ROI']=(orders_LTV_CAC['LTV']/orders_LTV_CAC['cac']-1)*100
orders_LTV_CAC=orders_LTV_CAC.sort_values(by='cohort_month')

display(orders_LTV_CAC)
````

Расчитаем классическую поронку основных событий.

Оптимальная последовательность событий для воронки:
- page_view — первый контакт пользователя с платформой (верхний уровень воронки).
- product_view — проявление интереса к конкретному товару.
- add_to_cart — явный сигнал намерения купить.
- checkout_start — начало оформления заказа.
- checkout_complete — финальный этап (но ещё не факт покупки — зависит от оплаты).
  
Эта воронка отражает естественный путь пользователя от знакомства с платформой до попытки совершить покупку.

````python
# Задаем порядок шагов в воронке
funnel_steps = ['page_view', 'product_view', 'add_to_cart', 'checkout_start', 'checkout_complete']
results = []
for channel in pa_marketplace_events_users['acq_channel'].unique():
    channel_df = pa_marketplace_events_users[pa_marketplace_events_users['acq_channel'] == channel]
    # Фильтруем датасет, оставляя только нужные шаги воронки
    funnel_df_steps = channel_df[channel_df['event_type'].isin(funnel_steps)]
    # Считаем число уникальных пользователей на каждом шаге
    funnel_counts = funnel_df_steps.groupby('event_type')['user_id'].nunique().reindex(funnel_steps, fill_value=0)

    for step, users in funnel_counts.items():
        results.append({
            'step': step,
            'users_control': users,
            'acq_channel': channel
        })

# Преобразуем список словарей в DataFrame
results_df = pd.DataFrame(results)

# Рассчитываем конверсии после создания DataFrame
# Конверсия от первого шага воронки
first_step_users = results_df.groupby('acq_channel')['users_control'].transform('first')
results_df['conversion_from_first_%'] = round(results_df['users_control'] / first_step_users * 100, 1)


# Визуализируем данные
plt.figure(figsize=(12, 12))
# Строим barplot (горизонтальный)
bars = sns.barplot(
    data=results_df,
    y='step',           # Этапы воронки на вертикальной оси
    x='conversion_from_first_%',  # Конверсия на горизонтальной оси
    hue='acq_channel'   # Разбивка по каналам привлечения
)

# Добавляем подписи для каждого столбца (справа от столбцов)
for bar_container in bars.containers:
    for bar_instance in bar_container:
        width = bar_instance.get_width()  # Длина столбца
        y_pos = bar_instance.get_y() + bar_instance.get_height() / 2  # Центр столбца по вертикали


        # Проверяем, что ширина не NaN
        if not pd.isna(width):
            plt.text(
                width + 0.3,  # Сдвигаем текст вправо от столбца (на 0.3 единицы)
                y_pos,  # Позиция по вертикали — центр столбца
                f'{width:.1f}%',
                ha='left',      # Выравнивание текста слева (т. е. справа от столбца)
                va='center',   # Выравнивание по центру вертикали
                fontsize=9
            )

plt.title('Классическая воронка: конверсия от первого шага')
plt.xlabel('Конверсия от первого шага (%)')  # Ось X — значения конверсии
plt.ylabel('Этапы воронки')  # Ось Y — этапы воронки
plt.legend(title='Канал привлечения')
plt.tight_layout()  # Улучшает отображение, предотвращает обрезание элементов
plt.show() 
````

Основные потери пользователей происходят на переходах между просмотром товара и добавлением товара в корзину, а также между корзиной и оформлением заказа. При этом структура воронки у разных каналов привлечения отличается незначительно, кроме пользователей по каналу привлечения TikTok, очень много пользователей отсеивается на всех шагах воронки.

Посморим сколько рублей прибыли прибыли приносит каждый рубль рекламы.

````python
# Общая комиссия (выручка)
revenue = (
    pa_marketplace_orders_users
    .groupby('acq_channel')['comission']
    .sum()
    .reset_index(name='Revenue')
)

# ROMI
marketing = (
    pa_marketplace_campaign_costs
    .groupby('acq_channel')['budget']
    .sum()
    .reset_index(name='Marketing_cost')
)

romi = revenue.merge(marketing, on='acq_channel')
romi['ROMI_%'] = (romi['Revenue'] - romi['Marketing_cost']) / romi['Marketing_cost'] * 100



plt.figure(figsize=(12,8))

plt.bar(
    romi['acq_channel'],
    romi['ROMI_%']
)

plt.title('ROMI по каналам привлечения')
plt.xlabel('Канал')
plt.ylabel('ROMI (%)')
plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
````

Affilate остаётся наиболее эффективным рекламным каналом с точки зрения возврата маркетинговых инвестиций. Google Ads за каждый рубль рекламы приносит прибыль. Несмотря на рост числа новых пользователей из TikTok, эффективность вложений в этот канал остаётся отрицательной.

````python
# Количество привлеченных пользователей
users = (
    pa_marketplace_users
    .groupby('acq_channel')['user_id']
    .nunique()
    .reset_index(name='users')
)

# CAC
cac = (
    profiles
    .groupby('acq_channel')['cac']
    .mean()
    .reset_index(name='CAC')
)

# Средний итоговый LTV
ltv = (
    ltv_month
    .groupby('acq_channel')['LTV']
    .mean()
    .reset_index(name='LTV')
)

# Средний итоговый ROI
roi = (
    ltv_month
    .groupby('acq_channel')['ROI']
    .mean()
    .reset_index(name='ROI')
)

# Средний чек
aov = (
    orders_channel
    .groupby('acq_channel')['total_price']
    .mean()
    .reset_index(name='AOV')
)

# Среднее количество покупок
orders_user = (
    (orders_channel.groupby('acq_channel')['order_id'].nunique() /
     orders_channel.groupby('acq_channel')['user_id'].nunique())
    .reset_index(name='Orders_per_user')
)

# Общая комиссия (выручка)
revenue = (
    pa_marketplace_orders_users
    .groupby('acq_channel')['comission']
    .sum()
    .reset_index(name='Revenue')
)

# Платящие пользователи
buyers = (
    pa_marketplace_orders_users
    .groupby('acq_channel')['user_id']
    .nunique()
    .reset_index(name='Buyers')
)

# Конверсия в покупку
buyers = buyers.merge(users, on='acq_channel')
buyers['Conversion_%'] = buyers['Buyers'] / buyers['users'] * 100

# ROMI
marketing = (
    pa_marketplace_campaign_costs
    .groupby('acq_channel')['budget']
    .sum()
    .reset_index(name='Marketing_cost')
)

romi = revenue.merge(marketing, on='acq_channel')
romi['ROMI_%'] = (romi['Revenue'] - romi['Marketing_cost']) / romi['Marketing_cost'] * 100

# Собираем всё вместе
summary = (
    users
    .merge(cac, on='acq_channel')
    .merge(ltv, on='acq_channel')
    .merge(roi, on='acq_channel')
    .merge(aov, on='acq_channel')
    .merge(orders_user, on='acq_channel')
    .merge(revenue, on='acq_channel')
    .merge(buyers[['acq_channel','Conversion_%']], on='acq_channel')
    .merge(romi[['acq_channel','ROMI_%']], on='acq_channel')
)

summary = summary.round(2)

display(summary)
````

#### Контекст и цель эксперимента
- **Бизнес-контекст:** Команда продукта запускает эксперимент, направленный на повышение конверсии среди пользователей, привлечённых через канал TikTok. Для стимулирования первой покупки пользователям будет предоставлен бонус на первую покупку.

---
#### Формат эксперимента
- **Тип эксперимента:** Сплит-роллинг (rolling split)

- **Период проведения:** 3 недели (фиксировано)

````python
# Нормализуем даты (оставляем только дату, время = 00:00:00)
users['registration_date'] = users['registration_date'].dt.normalize()
orders['order_date'] = orders['order_date'].dt.normalize()

first_orders=orders.sort_values('order_date').drop_duplicates('user_id', keep='first')
count_registr=users.groupby(['registration_date', 'group'])['user_id'].nunique().reset_index()
count_orders=first_orders.groupby(['order_date', 'group'])['user_id'].nunique().reset_index()

user_conv = count_orders.merge(count_registr, left_on=['order_date', 'group'], right_on=['registration_date', 'group'] , how='outer')
# Конверсия = покупатели / общее число × 100%
user_conv['convertion']=((user_conv['user_id_x'] /user_conv['user_id_y']) * 100).fillna(0)

 # Визуализация всех метрик на одном графике
plt.figure(figsize=(14, 8))
# Цикл для построения графиков по каналам
for group in user_conv['group'].unique():
    df_plot_user = user_conv[user_conv['group'] == group]
    plt.plot(df_plot_user['registration_date'], df_plot_user['convertion'], marker='o', label=f'{group}')


plt.title('Конверсию в покупку')
plt.xlabel('Дата')
plt.ylabel('Конверсия %')
plt.legend()
plt.grid(True)
plt.show()
````

Проанализируем основную метрику: конверсия в первую покупку - доля пользователей, зарегистрировавшихся с TikTok и совершивших первую покупку

Проверим гипотезу:

- H0: конверсии в тестовой и контролькой группах равны.
- H1: конверсии в тестовой и контрольной группах различаются.

````python
# Количество пользователей в каждой группе
users_count = users.groupby('group')['user_id'].nunique().reset_index(name='users')

# Количество пользователей, совершивших хотя бы одну покупку
buyers_count = orders.groupby('group')['user_id'].nunique().reset_index(name='buyers')

# Объединяем
conversion = users_count.merge(buyers_count, on='group', how='left')

conversion['buyers'] = conversion['buyers'].fillna(0).astype(int)

# Рассчитываем конверсию
conversion['conversion'] = conversion['buyers'] / conversion['users']

success = [
    conversion.loc[conversion['group']=='control','buyers'].iloc[0],
    conversion.loc[conversion['group']=='treatment','buyers'].iloc[0]
]

nobs = [
    conversion.loc[conversion['group']=='control','users'].iloc[0],
    conversion.loc[conversion['group']=='treatment','users'].iloc[0]
]

z_stat, p_value = proportions_ztest(success, nobs)

# Расчет лифта
conv_A = conversion.loc[conversion['group'] == 'control', 'conversion'].iloc[0]
conv_B = conversion.loc[conversion['group'] == 'treatment', 'conversion'].iloc[0]

results_df = pd.DataFrame({
    'Group A': ['control'],
    'Group B': ['treatment'],
    'Users A': [conversion.loc[conversion['group']=='control','users'].iloc[0]],
    'Users B': [conversion.loc[conversion['group']=='treatment','users'].iloc[0]],
    'Buyers A': [conversion.loc[conversion['group']=='control','buyers'].iloc[0]],
    'Buyers B': [conversion.loc[conversion['group']=='treatment','buyers'].iloc[0]],
    'Conversion A (%)': [conv_A * 100],
    'Conversion B (%)': [conv_B * 100],
    'Absolute Lift (п.п.)': [(conv_B - conv_A) * 100],
    'Relative Lift (%)': [(conv_B / conv_A - 1) * 100],
    'Z-statistic': [z_stat],
    'p-value': [p_value]
})

display(results_df)
````

По результатам статистического теста нулевая гипотеза о равенстве конверсий отвергается (p-value < 0.05). Разница между группами статистически значима. Предоставление бонуса на первую покупку положительно влияет на конверсию пользователей TikTok в первую покупку. В тестовой группе конверсия выросла с 1,06% до 7,51%, что соответствует увеличению более чем в 6 раз относительно контрольной группы. Таким образом, основная гипотеза эксперимента подтверждена: бонус на первую покупку стимулирует пользователей TikTok совершать первую покупку.

Проанализируем вспомогательную метрику: ARPPU (средняя выручка на покупателя TikTok)

Проверим гипотезу:

- H0: средняя выручка на покупателя  в тестовой и контролькой группах равны.
- H1: средняя выручка на покупателя  в тестовой и контрольной группах различаются.


````python
# выручка пользователя только среди покупателей

user_arppu = (
    orders
    .groupby(['user_id','group'])['comission']
    .sum()
    .reset_index()
)

control = user_arppu[
    user_arppu['group']=='control'
]['comission']

treatment = user_arppu[
    user_arppu['group']=='treatment'
]['comission']

from scipy.stats import ttest_ind

stat, p_value = ttest_ind(
    control,
    treatment,
    equal_var=False
)

arppu_result = pd.DataFrame({
    'Group A': 'control',
    'Group B': 'treatment',
    'p-value': p_value,
    'Mean A': [control.mean()],
    'Mean B': [treatment.mean()],
    'Absolute Lift': [treatment.mean()-control.mean()],
    'Relative Lift (%)': [(treatment.mean()/control.mean()-1)*100]
})

display(arppu_result)
````

По результатам статистического теста различия между группами являются статистически значимыми на уровне значимости 5%. Пользователи, получившие бонус, не только чаще совершали первую покупку, но и среди покупателей приносили больше выручки. Средняя выручка на одного платящего пользователя увеличилась почти на 48%. Это может свидетельствовать о том, что бонус дополнительно мотивировал пользователей совершать более ценные покупки.

Проанализируем вспомогательную метрику: ARPU (средняя выручка на пользователя TikTok)


Проверим гипотезу:

- H0: средняя выручка на пользователя в тестовой и контролькой группах равны.
- H1: средняя выручка на пользователя в тестовой и контрольной группах различаются.

````python
# выручка каждого пользователя

user_arpu = (
    users[['user_id','group']]
    .merge(
        orders.groupby('user_id')['comission'].sum(),
        on='user_id',
        how='left'
    )
)

user_arpu['comission'] = user_arpu['comission'].fillna(0)

control = user_arpu[
    user_arpu['group']=='control'
]['comission']

treatment = user_arpu[
    user_arpu['group']=='treatment'
]['comission']

stat, p_value = ttest_ind(
    control,
    treatment,
    equal_var=False
)

arpu_result = pd.DataFrame({
    'Group A': 'control',
    'Group B': 'group2',
    'p-value': p_value,
    'Mean A': [control.mean()],
    'Mean B': [treatment.mean()],
    'Absolute Lift': [treatment.mean()-control.mean()],
    'Relative Lift (%)': [(treatment.mean()/control.mean()-1)*100]
})

display(arpu_result)
````

Разница между группами статистически значима (p-value < 0.05). В тестовой группе средняя выручка на пользователя увеличилась более чем в 10 раз относительно контрольной группы. Рост ARPU связан прежде всего с существенным увеличением конверсии в первую покупку: благодаря бонусу значительно больше пользователей начали приносить выручку. Таким образом, эксперимент положительно влияет не только на поведение пользователей, но и на финансовый результат.

Проанализируем вспомогательную метрику: AOV (средний чек TikTok)

Проверим гипотезу:

Проанализируем барьерную метрику: Общая выручка от пользователей TikTok

Проверим гипотезу:

- H0: Общая выручка в тестовой и контролькой группах равны.
- H1: Общая выручка в тестовой и контрольной группах различаются.

````python
# Общая выручка по пользователям (включая пользователей без покупок)

user_revenue = (
    users[['user_id', 'group']]
    .merge(
        orders.groupby('user_id')['comission'].sum(),
        on='user_id',
        how='left'
    )
)

# пользователи без покупок имеют 0 выручки
user_revenue['comission'] = user_revenue['comission'].fillna(0)

# Разделяем группы
control = user_revenue[
    user_revenue['group'] == 'control'
]['comission']

treatment = user_revenue[
    user_revenue['group'] == 'treatment'
]['comission']

# Статистический тест
stat, p_value = ttest_ind(
    control,
    treatment,
    equal_var=False
)

# Расчет изменений
mean_A = control.mean()
mean_B = treatment.mean()

revenue_result = pd.DataFrame({
    'Group A': 'control',
    'Group B': 'treatment',
    'p-value': p_value,
    'Mean A': [control.mean()],
    'Mean B': [treatment.mean()],
    'Absolute Lift': [treatment.mean()-control.mean()],
    'Relative Lift (%)': [(treatment.mean()/control.mean()-1)*100]
})

display(revenue_result)
````
